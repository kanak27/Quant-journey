```python
import pandas as pd
import yfinance as yfin
import datetime as dt
import numpy as np
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

```python
Ticker = ['^NSEI']

# Indicate the start and end dates
start = dt.datetime.now().replace(year=dt.datetime.now().year - 1)
end = dt.datetime.now()

nifty = yfin.download(Ticker, start = start, end = end)
print(nifty)
```

```python
dailyReturns = nifty['Close'].pct_change(1) * 100
dailyReturns = dailyReturns.dropna()
```

```python
fig = plot_acf(dailyReturns, lags = 30)  # assign to avoid the duplicate-figure render
```

```python
fig = plot_pacf(dailyReturns, lags = 30, method = 'ywm')  # ywm avoids the FutureWarning
```
