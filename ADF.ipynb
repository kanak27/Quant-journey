```python
import pandas as pd
import yfinance as yfin
import datetime as dt
import numpy as np
from statsmodels.tsa.stattools import adfuller
```

```python
Ticker = ['^NSEI']

# Indicate the start and end dates
start = dt.datetime.now().replace(year=dt.datetime.now().year - 1)
end = dt.datetime.now()

nifty = yfin.download(Ticker, start = start, end = end)
print(nifty)
```

```python
closingPrices = nifty['Close']
closingPrices
```

```python
dailyReturns = nifty['Close'].pct_change(1) * 100
dailyReturns = dailyReturns.dropna()
```

```python
closingAdfResults = adfuller(closingPrices)

print(f"ADF Statistic: {closingAdfResults[0]}")
print(f"p-value: {closingAdfResults[1]}")
print("Critical Values:")
for key, value in closingAdfResults[4].items():
    print(f"   {key}: {value}")
```

```python
ReturnsAdfResults = adfuller(dailyReturns)

print(f"ADF Statistic: {ReturnsAdfResults[0]}")
print(f"p-value: {ReturnsAdfResults[1]}")
print("Critical Values:")
for key, value in ReturnsAdfResults[4].items():
    print(f"   {key}: {value}")
```
